```r
NycAvi.info <- read_tsv('/home/panda2bat/Avivorous_bat/output/07_genome-repeat/N.aviator/NycAvi.info', col_names = c('chromosome', 'end'))
```

```r
NycAvi.info <- NycAvi.info %>% filter(end >= 1000*1000) %>% mutate(start = 0) %>% select(c('chromosome', 'start', 'end')) %>% filter(end >= 1000*1000)
NycAvi.info
```

```r
info <- '/home/panda2bat/Avivorous_bat/output/07_genome-repeat/N.aviator/NycAvi.info'
gff <- '/home/panda2bat/Avivorous_bat/output/08_genome-annotation/N.aviator/PASA_update/pasa_align_jTnyB.gene_structures_post_PASA_updates.2548982.gff3'
circlize.gff <- circlize.prepare.gff(info, gff, feature = 'CDS', resolution = 1000*1000)
```

```r
edta <- '/home/panda2bat/Avivorous_bat/output/07_genome-repeat/N.aviator/N.aivtor.final.assembly.fa.mod.EDTA.TEanno.gff3'
circlize.edta <- circlize.prepare.gff(info, edta, resolution = 1000*1000)
```

```r
gap = '/home/panda2bat/Avivorous_bat/output/05_genome-scaffold/N.aviator/out_JBAT.FINAL.agp'
circlize.scaffold <- circlize.prepare.scaffold(gap)
circlize.scaffold
```

```r
col.lst <- RColorBrewer::brewer.pal(9, 'Set1')
circos.clear()
circos.par(start.degree = 90, gap.degree = c(rep(1, 19), 18))
circos.genomicInitialize(NycAvi.info, plotType = NULL)

#绘制 scaffold 图
circos.genomicTrack(
    circlize.scaffold, ylim = c(0, 1),
    panel.fun = function(region, value, ...){
        circos.genomicRect(
            region, value, ytop = 1, ybotom = 0,
            col = value$value2, border = NA
        )
    },
    track.height = .05,
    bg.border = NA,
    cell.padding = c(0, 1, .01, 1)
)
#绘制 cds 的占比折线图
circos.genomicTrack(
    circlize.gff, 
    panel.fun = function(region, value, ...) {
    circos.genomicLines(
      region, value, pch = 16, 
      cex = .1, col = col.lst[1], border = NA, area = TRUE, baseline = 'bottom', )
    },
    track.height = .15,
    bg.border = NA,
    cell.padding = c(0,1,.01,1),
)
circos.yaxis(sector.index = 'scaffold_1', track.index = 2, side = 'left', labels = paste0(c(round(CELL_META$ylim[1], 2), round(CELL_META$ylim[2],2))*100, '%'), at =c(round(CELL_META$ylim[1], 2), round(CELL_META$ylim[2], 2)), labels.cex = .5)
# 绘制重复序列的占比折线图
circos.genomicTrack(
    circlize.edta,
    panel.fun = function(region, value, ...) {
        circos.genomicLines(
            region, value, pch = 16,
            cex = .1, col = col.lst[2], border = NA, area = TRUE, baseline = 'bottom'
        )
    },
    track.height = .15,
    bg.border = NA,
    cell.padding = c(0,1,.01,1)
)
circos.yaxis(sector.index = 'scaffold_1', track.index = 3, side = 'left', labels.col = 'black', at = c(round(CELL_META$ylim[1],2), round(CELL_META$ylim[2],2)), labels.cex = .5, labels = paste0(c(round(CELL_META$ylim[1],2), round(CELL_META$ylim[2],2))*100, '%'))
# 绘制 gc 含量折线图
circos.genomicTrack(
    circlize.gc,
    panel.fun = function(region, value, ...){
        circos.genomicLines(
            region, value, pch = 16,
            cex = .1, col = col.lst[3], border = NA, area = TRUE, baseline = 'bottom'
        )
    },
    track.height = .15,
    bg.border = NA,
    cell.padding = c(0,1,.01,1)
)
# 绘制染色体的条带图
circos.yaxis(sector.index = 'scaffold_1', track.index = 4, side = 'left', labels.col = 'black', at = c(round(CELL_META$ylim[1],2), 0.66), labels.cex = .5, labels = paste0(c(round(CELL_META$ylim[1],2), 0.66)*100, '%'))

ideogram.col <- c('#a6cee3','#1f78b4','#b2df8a','#33a02c','#fb9a99','#e31a1c','#fdbf6f','#ff7f00','#cab2d6','#6a3d9a','#ffff99','#b15928', '#8dd3c7','#bebada','#fdb462','#fccde5','#bc80bd','#984ea3','#7fc97f', '#f0027f')
circos.track(
    ylim = c(0, 1),
    panel.fun = function(x,y) {
        chr = stringr::str_remove(CELL_META$sector.index, 'scaffold_')
        xlim = CELL_META$xlim
        ylim = CELL_META$ylim
        num = CELL_META$sector.numeric.index
        circos.rect(xlim[1], 0, xlim[2], 1, col = ideogram.col[num], border = '#d9d9d9')
        if((num<=17) || (num == 19)){
            circos.text(mean(xlim), mean(ylim) - mm_y(5), chr, cex = .7, col = 'black', facing = "bending.inside", niceFacing = TRUE)
        }
    },
    track.height = .05,
    bg.border = NA,
    cell.padding = c(0,1,.01,1)
)
```
