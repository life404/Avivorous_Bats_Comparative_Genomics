```python
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
pantherScore_lst = list(
    Path(
        "/home/panda2bat/Avivorous_bat/output/13_evolution-family_analysis/orthofinder/Results_Jul06/Phylogenetic_Hierarchical_Orthogroups/Hierarchical_Orthogroups_Sequences"
    ).glob("*.panther")
)
```

```python
# read the panther family information file
names_tab = {}
with open(
    "/home/panda2bat/TOOLS/PANTHER_HMMs/target/famlib/rel/PANTHER17.0_altVersion/hmmscoring/PANTHER17.0/globals/names.tab",
    "r",
) as file:
    for line in file:
        line = line.strip()
        array = line.split("\t")
        if array[0].endswith(".mag.mod"):
            names_tab[array[0].replace(".mag.mod", "")] = array[1]
```

```python
def parse_pantherScore(pantherScore_file, names_tab):
    try:
        df = pd.read_csv(pantherScore_file, sep="\t", header=None)
        df_annotation = df.iloc[:, 1]
        panther_family = [i.split(":")[0] for i in df_annotation]
        panther_family_dict = {i: panther_family.count(i) for i in set(panther_family)}
        max_panther = max(panther_family_dict, key=panther_family_dict.get)
        return (pantherScore_file.stem, max_panther, names_tab[max_panther])
    except pd.errors.EmptyDataError:
        return (pantherScore_file.stem, "null", "null")
```

```python
pantherScore_dict = {"HOG": [], "PANTHER": [], "DESCRIPTION": []}

for pantherScore_file in pantherScore_lst:
    hog, panther, description = parse_pantherScore(pantherScore_file, names_tab)
    pantherScore_dict["HOG"].append(hog)
    pantherScore_dict["PANTHER"].append(panther)
    pantherScore_dict["DESCRIPTION"].append(description)

pantherScore_df = pd.DataFrame(pantherScore_dict)
```
